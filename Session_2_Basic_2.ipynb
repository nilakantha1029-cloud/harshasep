**task**

- separate cat and num columns in different list
- using dtypes

```python
# idea
# extract keys and values
# apply the for loop
# check the condition if the data type = object then 
# append in cat list
# other wise append in num list
```

```python
Visa_df 
Visa_df.dtypes
char=[]
num=[]
keys=Visa_df.dtypes.keys()
val=Visa_df.dtypes.values
for i,j in (zip(keys,val)):
    if j == 'object':
        char.append(i)
    else:
        num.append(i)
print(char)
print(num)
```
