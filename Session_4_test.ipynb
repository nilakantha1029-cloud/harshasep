```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
visa_df=pd.read_csv(r"C:\Users\DELL\Naresh It\Data_files\Visadataset.csv")
```

```python
visa_df
```

```python
keys=visa_df.dtypes.keys()
values=visa_df.dtypes.values
keys,values
```

```python
cat=[]
num=[]
for i,j in zip(keys,values):
    if  j=='object':
        cat.append(i)
    else:
        num.append(i)
cat,num
```

```python
import os 
current_dir=os.getcwd()
os.makedirs('Frequency_Table',exist_ok=True)
new_folder='Frequency_Table'
new_dir=os.path.join(current_dir,new_folder)
# new_dir
for i in cat[1:]:
    colpath=f"{i}.csv"
    finalpath=os.path.join(new_dir,colpath)
    keys=visa_df[i].value_counts().keys()
    values=visa_df[i].value_counts().values
    df=pd.DataFrame(zip(keys,values))
    df.to_csv(finalpath)
```

```python
plt.figure(figsize=(22, 14))
for i,j in enumerate(cat[1:]):

    keys=visa_df[j].value_counts().keys()
    values=visa_df[j].value_counts().values
    
    plt.subplot(3,3,i+1)
    plt.bar(keys,values)
    plt.title(f"{j} bar_chat")
    plt.savefig(f"C:\\Users\\DELL\\Naresh It\\EDA\\bar_charts\\{j}.jpg")
plt.show()
```

```python
for i in cat[1:]:
    keys=visa_df[i].value_counts().keys()
    values=visa_df[i].value_counts().values
    plt.figure(figsize=(10, 6))
    plt.pie(values,labels=keys)
    plt.xlabel("Lables")
    plt.ylabel("Count")
    plt.title(f"{i} pie_chat")
    plt.savefig(f"C:\\Users\\DELL\\Naresh It\\EDA\\pie_charts\\{j}.jpg")
plt.show()
```

```python
import os
current_dir=os.getcwd()
os.makedirs("Bar_Chart",exist_ok=True)
new_fol="Bar_Chart"
newdir=os.path.join(current_dir,new_fol)
plt.figure(figsize=(22,14))
for i,j in enumerate(cat[1:]):
    filetp=f"{j}.jpg"
    finalpath=os.path.join(newdir,filetp)
    keys=visa_data[j].value_counts().keys()
    values=visa_data[j].value_counts().values
    plt.subplot(3,3,i+1)
    plt.bar(keys,values)
    plt.title(f"{j} bar_chart")
    plt.xlabel(f"{j}")
    plt.ylabel("Counts")
    plt.savefig(finalpath)
plt.show()
```
