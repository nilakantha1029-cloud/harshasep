**Step-12: Append a row**

**loc-iloc**

```python
df.loc[rows,columns]
    rows are index numbers similar to list slice
    l[start:stop:step]
    column are names ant column name we can provide 
df.iloc[rows,columns]
    rows and columns are index numbers similar to list slice
    l[start:stop:step]
```

```python
rows=[1,2,3]
cols=['Age','City']
df.loc[rows,cols]
```

```python
rows=[1,2,3]
cols=[1,2]
df.iloc[rows,cols]
```

```python
cols=['Age','City']
df.loc[1:3,cols]
```

```python
cols=[1,2]
df.iloc[1:4,1:3]
```

```python
df.loc[2]
```
