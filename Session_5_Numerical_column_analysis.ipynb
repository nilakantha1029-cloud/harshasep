```python
def emperical(num,alpha):
    wage_df=visa_df['prevailing_wage']
    Mean=np.mean(wage_df)
    Mean=np.mean(wage_df)
    Sd=np.std(wage_df)
    lb=Mean-alpha*Sd
    ub=Mean+alpha*Sd
    con1=wage_df>lb
    con2=wage_df<ub
    con=con1 & con2
    return np.sum(con)==num*(25480)/100
emperical(68,1),emperical(95,1),emperical(99.7,1)
```
